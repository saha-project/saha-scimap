```python
import pandas as pd
import numpy as np
import anndata as ad
import scanpy as sc
import matplotlib.pyplot as plt
```

```python
PRT_ALL = sc.read('../SAHAMaxFuse/SAHA_All_PRT_share.h5ad')
```

```python
# Stomach
PRT = sc.read('PRT_STO_SM.h5ad')
PRT_name = 'SAHA_CP_STO_A1_EC03'
PRT = PRT[PRT.obs['SAHA_name']==PRT_name].copy()
PRT.obs['fov'] = PRT_ALL[PRT_ALL.obs['SAHA_name']==PRT_name].obs['fov']

PRT.obs['phenotype'] = PRT.obs['phenotype'].str.replace('likely-', '').astype('category')

# Define your mapping from fine to broad phenotype
phenotype_mapping = {
    'Stroma': 'Mesenchymal-Stromal',
    'Endothelial': 'Endothelial',
    'Epithelial': 'Epithelial',
    'Myeloid': 'Myeloid',
    'Immune': 'Lymphoid-Other',
    'Dendritic Cells': 'Lymphoid-DC',
    'T Cells': 'Lymphoid-T',
    'B Cells': 'Lymphoid-B',
}

# Apply the mapping
PRT.obs['phenotype_broad'] = PRT.obs['phenotype'].map(phenotype_mapping)
```

```python
PRT.obs['fov_int'] = pd.to_numeric(PRT.obs['fov'], errors='coerce')
fovs = [4,5,6,7,11,12,13,14,19,20,21,22,27,28,29,30]
size = 35

palette_PRT = [
    '#ff99cc', #B
    '#7DD520', #DC
    '#ff7f0e', #endo
    '#1f77b4', #epi
    '#17becf', #Other
    '#d62728', #Myeloid
    '#2ca02c', #Mesen
    '#8c564b', #T
]

fig, ax = plt.subplots(1, 1, figsize=(10, 10))

sc.pl.embedding(PRT[PRT.obs['fov_int'].isin(fovs)], "spatial", color='phenotype_broad', ax=ax, s=size, show=False, palette=palette_PRT)
# sc.pl.embedding(PRT[PRT.obs['fov_int'] <= 65], "spatial", color='celltype_mf', ax=ax, s=15, show=False)

# Customize the plot
ax.set_xlabel('')
ax.set_ylabel('')
ax.set_aspect('equal')
ax.set_facecolor('white')
ax.set_title(PRT_name)

# Remove the spines (borders)
for spine in ax.spines.values():
    spine.set_visible(False)

plt.tight_layout()
plt.savefig("STO_PRT.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
